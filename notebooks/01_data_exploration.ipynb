```python
# PHASE 4 — Decide how we represent a Natura 2000 site
# STEP 1:
natura_site = (
    natura[["nr", "geometry"]] # gets the nr and geometry column from the natura dataset, because we want natura_site to nly contain these 2 columns
    .dissolve(by="nr", as_index=False)
)
```

```python
print(natura_site.columns) # ensuring we have exactly those 2 columns
display(natura_site.head())
```

```python
# STEP 2:
site_names = (
    natura.groupby("nr")["naam_n2k"]
    .first()
    .reset_index()
)
display(site_names.head())
print(site_names.columns)
```

```python
# Now we have 2 tables:
# natura_site and site_names
# the naura_siet table contains the geometry column having the geometries for each nr:
#     nr    geometry
#     --------------
#     1     Geometry A
#     2     Geometry B
#     3     Geometry C

# whereas,
# site_names contains the naam_n2k column having the site name associated to each nr:
#     nr    naam_n2k
#     --------------
#     1     Waddenzee
#     2     Texel
#     3     Vlieland
# So now we are going to merge them using the merge()
# both tables have nr in common so now we match the rows where nr is the same in both the tables
```

```python
# STEP 3:
natura_site = natura_site.merge(
    site_names,
    on="nr", # Using the nr column as the matching key
    how="left" # this means to keep all rows from the left table into the new merged table. Here the left table is natura_site, and it has a total of 162 rows, so the merged table will also have 162 rows
)
```

```python
print(natura_site.columns)
```

```python
display(natura_site.head())
```

```python
# STEP 4: Creating the designation_by_site table
designation_by_site = (
    natura.groupby("nr")["beschermin"] # we group all rows having the same nr values and for those grouped rows -> we select only the beschermin column
    .apply(lambda x: ", ".join(sorted(x.unique())))
    .reset_index(name="designation_types")
)

# we treate all the column values of the beschermin column as a continuous string of values -> store that string into x 
# we join the individual values of the x string with a comma, this way they become list values -> so for each nr -> the x avlues are stored like a list [VR, VR+HR]
# here lambda helped us temporarily define this procedure of the string to List conversion of x values
# we used the apply() to actually apply the above conversion procedure defined in lambda for x
# lastly we renamed the beschermin column to designation_types column

# This also answers the question:
# Q What designation categories are associated with each Natura 2000 site? -> its answer is the resulting designation_by_site table


```

```python
print(designation_by_site.columns)
```

```python
display(designation_by_site.head())
```

```python
# STEP 5: Merging the designation_by_site table with natura_site table:
natura_site = natura_site.merge(
    designation_by_site,
    on="nr",
    how="left"
)
```

```python
# So the gdf_2005 dataframe specifically contains the data for the 2005 layer.
# now we will select only the ID column from this 2005 dataframe
ids_2005 = set(gdf_2005["gdn_grid_cell_id"])# -> set() eliminates any duplicate column values from the dataframe -> and considers only unique values.
print(ids_2005)
# now ids_2005 contains all those unique IDs present in 2005 layer.
```

```python
# Loop through the 20 layers again:
for layer in layers:
    gdf = gpd.read_file(gpkg_path, layer = layer) # Loading the current layer of the RIVM dataset being iterated over
    ids = set(gdf["gdn_grid_cell_id"])  # Here we extract the ID of of the current layer in the ongoing iteration and store it in the ids variable

# Here we have made 2005 ID as our reference ID, to which we will compare the IDs of the remaining layers
    
    print(
        layer,
        "→ same IDs as 2005:",
        ids == ids_2005  # We compare the current ID in the ids variable with the id of the 2005 layer to see if they match. -> if YES = It returns True
    )
```

```python
# Analysis:
# Above we have seen that the IDs of all layers match with the ID of the 2005 layer
```

```python
# 4. Checking If an ID represents the exact same polygon in every year:

# We take the 2 columns from the dataframe gdf_2005 which stores the 2005 layer data -> those 2 columns are: "gdn_grid_cell_id", "geometry" -> then we create a copy of those 2 columns -> store that copy of the two in the "reference" dataframe

reference = gdf_2005[["gdn_grid_cell_id","geometry"].copy()]
# YOU CAN CHECK USING reference.head()
reference.head()
```

```python
for layer in layers:
    # Iteratinhg over the 20 layers of the RIVM dataset, one iteration at a time

# THIS IS WHAT HAPPENS ATE EACH ITERATION:

# ACTION 1--------------------

    # loading the current layer from the RIVM dataset located at gpkg_path
    # storing the loaded layer in the gdf dataframe -> now the gdf dataframe contains the data for that cuurent layer
    gdf = gpd.read_file(gpkg_path, layer = layer)

# ACTION 2-------------------
    # Now we fetch the 2 columns ["id", "Geometry"] from the gdf dataframe and create a copy of the 2 columns -> store that copy in a test named dataframe
    test = gdf[["gdn_grid_cell_id", "geometry"]].copy()
    # you can check using test.head()
   
```

```python
# ACTION 3------------------

# MERGING THE GEOMETRIES OF 2005 YEAR AND CURRENTLY PROCESSED YEAR SHARING THE SAME IDS
# Now we are going to match each year's geometry to the reference (2005) year's geometry
# Now we Match the 2005 reference geometry and the current year's geometry using the grid-cell ID as the matching KEY -> meaning there will be only one ID column which contains IDs identical to the geometries in the corresponding columns
# We use merge() to match the reference dataframe to the test dataframe:
merged = reference.merge(
    test,
    on = "gdn_grid_cell_id",
    suffixes=("_2005", "_test") # These suffixes will be added to the geometry column of the reference and test dataframes. like this: _2005 will be added to the geometry column of the reference dataframe and _test will be added to geometry column of the test dataframe, so that it becomes easier to understand which column belongs to which dataframe while both of the dataframes are merged
)
# Finally storing the merged dataframes in the new "merged"  named dataframe
# YOU CAN CHECK USING merged.head()
merged.head()
```

```python
# ACTION 4------------------

# Now we check if the geometry_2005 column and the geometry_test column of the merged dataframe are equal:
same_geometry = merged["geometry_2005"].equals(merged["geometry_test"])
```

```python
# ACTION 5------------------

print(
    layer,
    "-> geometries Identical: ",
    same_geometry
)
```

```python
# The check returned True -> meaning the geometries of the 2005 layer and all the other layers are equal!
# Now we have completed a meaningful validation milestone: we have established that the 20 yearly RIVM layers are structurally compatible and can be treated as one consistent spatiotemporal dataset.
```

```python
# STEP 6: COMBINING THE 20 YEARLY RIVM LAYERS
# Each of the layers is a GeoDataframe, so 20 layers means 20 GeoDataframe, each of the dataFrame is appended to the list as a whole

# At the moment, the data is split across 20 layers -> Each layer has the same structure -> So we are going to combine all the layers into one GeoDataFrame.
# So the new dataframe would contain -> 43,959 cells × 20 years = 879,180 rows


# ACTION 1----------------------
# Creating a list to hold all the layers of the the RIVM dataset yearly:
rivm_layers = [] # creating an empty list for now, to append layers to it in each iteration
for layer in layers:
    gdf = gpd.read_file(gpkg_path, layer = layer)  # In the 1st iteration -> taking the 1st layer's data from the RIVM dataset located at gpkg_path. This 1st layer is the "ndep_historisch_2005" -> storing this is the gdf dataframe
    # Now with one layer in -> the gdf actually conatinsa a small dataframe of the entire 2005 layer
    rivm_layers.append(gdf) # Appending that layer from the gdf dataframe into the rivm_layers list
    # Then we move to the next iteration -> here also the 2nd layer's data is loaded in a way that the layer is actually a small dataset of all the columns: gdn_grid_cell_id	substance_name	year	deposition	geometry especially for the 2006 layer

```

```python
# Now we have appended all the 20 layer dataframes to the rivm_layers list

# ACTION 2-------------------
# Now we check the length of the list to ensure we have loaded all the 20 layers Dataframes
print(len(rivm_layers))
```

```python
# ACTION 3-------------------
# Now we combine all of the dataframes stored in the rivm_layers dataframe using the concat() method of pandas -> storing the combined lists into the rivm_historical named dataframe
import pandas as pd
rivm_historical = pd.concat(rivm_layers, ignore_index=True) # The concat() vertically stacks the GeoDataframes stored in the list
# ignore_index=True -> Forget the old indexes and create a new continuous index right after the end of the 1st dataframe. Without ignore_index=True -> we'd get repeated indexes.
print(rivm_historical)
print(type(rivm_historical))
```
