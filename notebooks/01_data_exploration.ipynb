```python
# PHASE 4 — Decide how we represent a Natura 2000 site
# STEP 1:
natura_site = (
    natura[["nr", "geometry"]] # gets the nr and geometry column from the natura dataset, because we want natura_site to nly contain these 2 columns
    .dissolve(by="nr", as_index=False)
)
```

```python
print(natura_site.columns) # ensuring we have exactly those 2 columns
display(natura_site.head())
```

```python
# STEP 2:
site_names = (
    natura.groupby("nr")["naam_n2k"]
    .first()
    .reset_index()
)
display(site_names.head())
print(site_names.columns)
```

```python
# Now we have 2 tables:
# natura_site and site_names
# the naura_siet table contains the geometry column having the geometries for each nr:
#     nr    geometry
#     --------------
#     1     Geometry A
#     2     Geometry B
#     3     Geometry C

# whereas,
# site_names contains the naam_n2k column having the site name associated to each nr:
#     nr    naam_n2k
#     --------------
#     1     Waddenzee
#     2     Texel
#     3     Vlieland
# So now we are going to merge them using the merge()
# both tables have nr in common so now we match the rows where nr is the same in both the tables
```

```python
# STEP 3:
natura_site = natura_site.merge(
    site_names,
    on="nr", # Using the nr column as the matching key
    how="left" # this means to keep all rows from the left table into the new merged table. Here the left table is natura_site, and it has a total of 162 rows, so the merged table will also have 162 rows
)
```

```python
print(natura_site.columns)
```

```python
display(natura_site.head())
```

```python
# STEP 4: Creating the designation_by_site table
designation_by_site = (
    natura.groupby("nr")["beschermin"] # we group all rows having the same nr values and for those grouped rows -> we select only the beschermin column
    .apply(lambda x: ", ".join(sorted(x.unique())))
    .reset_index(name="designation_types")
)

# we treate all the column values of the beschermin column as a continuous string of values -> store that string into x 
# we join the individual values of the x string with a comma, this way they become list values -> so for each nr -> the x avlues are stored like a list [VR, VR+HR]
# here lambda helped us temporarily define this procedure of the string to List conversion of x values
# we used the apply() to actually apply the above conversion procedure defined in lambda for x
# lastly we renamed the beschermin column to designation_types column

# This also answers the question:
# Q What designation categories are associated with each Natura 2000 site? -> its answer is the resulting designation_by_site table


```

```python
print(designation_by_site.columns)
```

```python
display(designation_by_site.head())
```

```python
# STEP 5: Merging the designation_by_site table with natura_site table:
natura_site = natura_site.merge(
    designation_by_site,
    on="nr",
    how="left"
)
```

```python
print(natura_site.columns)
```

```python
display(natura_site.head(20))
```

```python
# STEP 6: CALCULATION OF AREA:
natura_site["site_area_m2"] = natura_site.geometry.area
# natura_site.geometry.area -> accesses the GeoPandas geometry column and .area calculates the geometry of each column -> we assign the result to "site_area_m2"
```

```python
# STEP 7: CONVERTING THE AREA FROM SQUARE METRES TO HECTARES
natura_site["site_area_ha"] = (
    natura_site["site_area_m2"] / 10_000
)
# 1 hectare = 10,000 m²
# Area in hectares = area in square metres / 10,000
```

```python
# we reorder the columns of natura_site into a proper order
natura_site = natura_site[
    [
        "nr",
        "naam_n2k",
        "designation_types",
        "geometry",
        "site_area_m2",
        "site_area_ha"
    ]
]
```
